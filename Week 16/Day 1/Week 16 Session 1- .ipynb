```python
#Question: Write a Python program that asks the user to enter a total number of days. The program should convert this input into the equivalent number of Years, Weeks, and Remaining Days.
```

```python
# Step 1: Get input from the user
total_days = int(input("Enter the number of days: ")) 

# Step 2: Calculate years using Integer Division (//)
# We use // to get the whole number of years
years = total_days // 365     # 375//365 = 1

# Step 3: Find remaining days after taking out the years using Remainder (%) الباقي 
remaining_days_after_years = total_days % 365   # 375%365 = 10

# Step 4: Calculate weeks from the remaining days
weeks = remaining_days_after_years // 7   # 10//7 = 1

# Step 5: Find the final remaining days using Remainder (%)
days = remaining_days_after_years % 7   # 10% 7 = 3
# Step 6: Print the result هنا بقوله اللي طلعه من الحسابات دي ايه 
print(f"\nResults for {total_days} days:")
print(f"Years: {years}")
print(f"Weeks: {weeks}")
print(f"Days: {days}")
```

1-The Area (Length × Width). قانو المساحه 

2-The Perimeter (2 × (Length + Width)). المحيط 

```python
l=float(input("Enter the length:"))# فلوتى علشان ممكن يكون في كسور عادي 
w=float(input("Enter the width:"))
area=l*w
perimeter=2*(l+w)
print(f'area={area}\nperimeter={perimeter}') # خلي بالكمن سلاش ان علشان يبقى في سطرين مش سطر واحد 
```

## حل الدكتور للسوال اللي فات الفرق بس ان الدكتور كاتبه تفاصيل 

```python
# 1. Get user input for length and width
# We use float() to allow decimal values for dimensions
length = float(input("Enter the length of the rectangle: "))
width = float(input("Enter the width of the rectangle: "))

# 2. Calculate Area and Perimeter
# Area formula: Length * Width
area = length * width
# Perimeter formula: 2 * (Length + Width)
perimeter = 2 * (length + width)
# 3. Create a formatted report
print(f"\n--- Rectangle Results ---")
print(f"Length: {length} | Width: {width}")
print(f"The Area is: {area:.2f}")
print(f"The Perimeter is: {perimeter:.2f}")
#2f يعني اظهر اول رقمين  بعد العلامه بس كده 
# 4. Using 'pass' if we want to add future checks 
if length == width: 
    print("Note: This rectangle is a square!")
else:
    # No special note needed for non-square rectangles
    pass
```

```python
print(f"\nSet A: {set_a}")
print(f"Set B: {set_b}")

print(f"Union (A | B): {set_a | set_b}")  # All elements
print(f"Intersection (A & B): {set_a & set_b}")  # Common elements
print(f"Difference (A - B): {set_a - set_b}")  # In A but not in B
print(f"Symmetric Difference (A ^ B): {set_a ^ set_b}")  # In A or B, but not both
```

# 2.3 While Loop

```python
# Function with Multiple Parameters
def calculate_bmi(weight_kg, height_m):
    """
    Calculate Body Mass Index (BMI)
    
    Parameters:
        weight_kg: Weight in kilograms
        height_m: Height in meters
        
    Returns:
        BMI value (float)
    """
    bmi = weight_kg / (height_m ** 2)
    return bmi
```

```python
bmi = calculate_bmi(70, 1.75)
print(f"\nBMI: {bmi}")
```

```python
# Function with Default Parameters
def power(base, exponent=2):
    """Calculate power with default exponent of 2 (square)"""
    return base ** exponent
```

```python
print(f"\n5^2 = {power(6)}")
print(f"5^3 = {power(5, 3)}")
```

```python
# Function with Multiple Return Values
def analyze_data(numbers):
    """Return statistics for a list of numbers"""
    total = sum(numbers)
    count = len(numbers)
    average = total / count
    minimum = min(numbers)
    maximum = max(numbers)
    return total, average, minimum, maximum

```

```python
data = [10, 20, 30, 40, 50]
total, avg, min_val, max_val = analyze_data(data)

print(f"\nData Analysis:")
print(f"Total: {total}, Average: {avg}, Min: {min_val}, Max: {max_val}")
```

## 4.2 Advanced Function Concepts

```python
# *args - Variable Number of Arguments
def calculate_average(*numbers):
    """Calculate average of any number of values"""
    if len(numbers) == 0:
        return 0
    return sum(numbers) / len(numbers)

print("Averages:")
print(f"Average of 10, 20: {calculate_average(10, 20)}")
print(f"Average of 10, 20, 30, 40: {calculate_average(10, 20, 30, 40)}")
```

```python
# **kwargs - Keyword Arguments
def create_student_profile(**info):
    """Create student profile from keyword arguments"""
    print("\nStudent Profile:")
    for key, value in info.items():
        print(f"  {key.capitalize()}: {value}")

create_student_profile(name="John", age=20, major="Data Science", gpa=3.8)
```
