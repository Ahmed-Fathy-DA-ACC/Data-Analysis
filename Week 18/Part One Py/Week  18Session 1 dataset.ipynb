## 7.6 Handling Missing Data

````python
# Create DataFrame with missing values
data_with_nulls = {
    'Product': ['A', 'B', 'C', 'D', 'E', 'F'],
    'Price': [10.5, np.nan, 15.0, 12.5, np.nan, 18.0],
    'Quantity': [100, 150, np.nan, 200, 175, 120],
    'Category': ['Electronics', 'Electronics', None, 'Clothing', 'Clothing', 'Electronics']
}

df_nulls = pd.DataFrame(data_with_nulls)
print("DataFrame with missing values:")
print(df_nulls)
````

````python
# Detect missing values
print("\nMissing values (True = missing):")
print(df_nulls.isnull())
````

````python
# Detect missing values
print("\nMissing values (True = missing):")
print(df_nulls.isnull().sum())
````

````python
# Fill missing values
df_filled = df_nulls.copy()

# Fill with specific value
df_filled['Price'] = df_filled['Price'].fillna(0)
print("\nAfter filling Price with 0:")
print(df_filled)
````

````python
# Fill with mean
df_filled2 = df_nulls.copy()
df_filled2['Price'] = df_filled2['Price'].fillna(df_filled2['Price'].mean())
df_filled2['Quantity'] = df_filled2['Quantity'].fillna(df_filled2['Quantity'].mean())
print("\nAfter filling with mean:")
print(df_filled2)
````

````python
# Forward fill (use previous value)
df_ffill = df_nulls.copy()
df_ffill['Price'] = df_ffill['Price'].fillna(method='ffill')
print("\nAfter forward fill:")
print(df_ffill)
````

````python
# Drop rows with any missing values
df_dropped = df_nulls.dropna()
print("\nAfter dropping rows with any NaN:")
print(df_dropped)
````

````python
# Drop rows where specific column is NaN
df_dropped_subset = df_nulls.dropna(subset=['Price'])
print("\nAfter dropping rows where Price is NaN:")
print(df_dropped_subset)
````

## 7.7 GroupBy and Aggregation

````python
import pandas as pd
import numpy as np

# Sales dataset
sales_data = {
    'Date': ['2024-01-01', '2024-01-01', '2024-01-02', '2024-01-02', '2024-01-03', '2024-01-03'],
    'Product': ['Laptop', 'Phone', 'Laptop', 'Tablet', 'Phone', 'Laptop'],
    'Category': ['Electronics', 'Electronics', 'Electronics', 'Electronics', 'Electronics', 'Electronics'],
    'Quantity': [2, 5, 3, 2, 4, 1],
    'Price': [1000, 800, 1000, 500, 800, 1000],
    'Region': ['North', 'South', 'North', 'East', 'South', 'West']
}

df_sales = pd.DataFrame(sales_data)
df_sales
````

````python
df_sales['Total'] = df_sales['Quantity'] * df_sales['Price']
df_sales
````

````python
print("Sales DataFrame:")
print(df_sales)
````

````python
# Group by single column
print("\n=== Group by Product ===")
grouped_product = df_sales.groupby('Product')['Total'].sum()
print(grouped_product)
````

````python
# Multiple aggregations
print("\n=== Multiple Aggregations by Product ===")
agg_result = df_sales.groupby('Product').agg({
    'Quantity': 'sum',
    'Total': ['sum', 'mean', 'count']
})
print(agg_result)
````

````python
# Group by multiple columns
print("\n=== Group by Product and Region ===")
multi_group = df_sales.groupby(['Product', 'Region'])['Quantity'].sum()
print(multi_group)
````

````python
# Reset index to make it a regular DataFrame
multi_group_df = df_sales.groupby(['Product', 'Region'])['Total'].sum().reset_index()
print("\nAs DataFrame:")
print(multi_group_df)
````

# 7.8 Sorting and Ranking

````python
# Create sample DataFrame
students_data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Math': [95, 87, 92, 78, 88],
    'Science': [88, 92, 85, 90, 95],
    'English': [92, 85, 88, 85, 90]
}

df_students = pd.DataFrame(students_data)
print("Students DataFrame:")
print(df_students)
````

````python
# Sort by single column
sorted_math = df_students.sort_values('Math', ascending=False)
print("\nSorted by Math (descending):")
print(sorted_math)
````

````python
# Sort by multiple columns
sorted_multi = df_students.sort_values(['Science', 'Math'], ascending=[False, False])
print("\nSorted by Science, then Math:")
print(sorted_multi)
````

````python
# Calculate total score
df_students['Total'] = df_students[['Math', 'Science', 'English']].sum(axis=1)
df_students['Average'] = df_students[['Math', 'Science', 'English']].mean(axis=1)
df_students
````

````python
# Ranking
df_students['Rank'] = df_students['Total'].rank(ascending=False, method='min')
df_students = df_students.sort_values('Rank')

print("\nFinal Results with Rankings:")
print(df_students)
````

## 💡 Real-World Project: Customer Sales Analysis
Let's put everything together in a comprehensive data analysis project!

````python
# Create comprehensive sales dataset
import pandas as pd
import numpy as np
np.random.seed(42)

dates = pd.date_range('2024-01-01', periods=100, freq='D')
customers = ['Customer_' + str(i) for i in range(1, 21)]
products = ['Laptop', 'Phone', 'Tablet', 'Headphones', 'Watch']
regions = ['North', 'South', 'East', 'West']

sales_comprehensive = {
    'Date': np.random.choice(dates, 200),
    'Customer': np.random.choice(customers, 200),
    'Product': np.random.choice(products, 200),
    'Region': np.random.choice(regions, 200),
    'Quantity': np.random.randint(1, 10, 200),
    'Unit_Price': np.random.choice([299, 799, 499, 149, 399], 200)
}
````

````python
df_comprehensive = pd.DataFrame(sales_comprehensive)
df_comprehensive.head()
````

````python
df_comprehensive.info()
````

````python
df_comprehensive.describe()
````

````python
df_comprehensive['Total_Sale'] = df_comprehensive['Quantity'] * df_comprehensive['Unit_Price']
df_comprehensive['Date'] = pd.to_datetime(df_comprehensive['Date'])
df_comprehensive['Month'] = df_comprehensive['Date'].dt.month
df_comprehensive['Day_of_Week'] = df_comprehensive['Date'].dt.day_name()
df_comprehensive
````

````python
num_data=df_comprehensive[["Quantity", "Unit_Price", "Total_Sale"]]
cor=num_data.corr()
import seaborn as sns
sns.heatmap(cor,annot=True)
````

````python
# عرض أبعاد مجموعة البيانات (شكل البيانات)
print(f"\nDataset shape: {df_comprehensive.shape}")

# تحليل رقم 1: أفضل 5 عملاء حسب الإيرادات
print("\n=== TOP 5 CUSTOMERS BY REVENUE ===")
# تجميع البيانات حسب العميل، جمع إجمالي المبيعات، ترتيبها تنازلياً، وأخذ أعلى 5
top_customers = df_comprehensive.groupby('Customer')['Total_Sale'].sum().sort_values(ascending=False).head(5)
print(top_customers)
````

````python
# Analysis 1: Top 5 customers by revenue
print("\n=== TOP 5 CUSTOMERS BY REVENUE ===")
top_customers = df_comprehensive.groupby('Customer')['Total_Sale'].sum().sort_values(ascending=False).head(5)
print(top_customers)
````

````python
# Analysis 2: Best-selling products
print("\n=== BEST-SELLING PRODUCTS ===")
product_analysis = df_comprehensive.groupby('Product').agg({
    'Quantity': 'sum',
    'Total_Sale': 'sum'
}).sort_values('Total_Sale', ascending=False)
print(product_analysis)
````

````python
# Analysis 4: Monthly trend
print("\n=== MONTHLY SALES TREND ===")
monthly = df_comprehensive.groupby('Month')['Total_Sale'].sum().sort_index()
print(monthly)
````

````python
# Analysis 5: Day of week analysis
print("\n=== SALES BY DAY OF WEEK ===")
# تجميع البيانات حسب اليوم وحساب الإجمالي، المتوسط، والعدد
day_analysis = df_comprehensive.groupby('Day_of_Week')['Total_Sale'].agg(['sum', 'mean', 'count'])
# تحديد ترتيب الأيام الصحيح
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
# إعادة ترتيب البيانات بناءً على الترتيب الأسبوعي
day_analysis = day_analysis.reindex([day for day in day_order if day in day_analysis.index])
print(day_analysis)
````

````python
# Summary statistics
print("\n=== OVERALL STATISTICS ===")
print(f"Total Revenue: ${df_comprehensive['Total_Sale'].sum():,.2f}")
print(f"Average Transaction: ${df_comprehensive['Total_Sale'].mean():,.2f}")
print(f"Total Transactions: {len(df_comprehensive)}")
print(f"Unique Customers: {df_comprehensive['Customer'].nunique()}")
````

## 8.5 Histograms
Histograms show the distribution of a single variable.

````python
# Generate sample data - exam scores
np.random.seed(42)
scores = np.random.normal(75, 10, 1000)  # mean=75, std=10

plt.figure(figsize=(12, 5))

# Histogram
plt.subplot(1, 2, 1)
plt.hist(scores, bins=30, color='#667eea', alpha=0.7, edgecolor='black')
plt.title('Distribution of Exam Scores', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Score', fontsize=11)
plt.ylabel('Frequency', fontsize=11)
plt.axvline(scores.mean(), color='red', linestyle='--', linewidth=2, label=f'Mean = {scores.mean():.1f}')
plt.legend()
plt.grid(axis='y', alpha=0.3)
# Histogram with density (normalized)
plt.subplot(1, 2, 2)
plt.hist(scores, bins=30, color='#764ba2', alpha=0.7, edgecolor='black', density=True)
plt.title('Normalized Distribution', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Score', fontsize=11)
plt.ylabel('Density', fontsize=11)
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
````

# 8.6 Pie Charts
Pie charts show proportions and percentages of a whole.

````python
# Market share data
labels = ['Python', 'Java', 'JavaScript', 'C#', 'Others']
sizes = [30, 25, 20, 15, 10]
colors = ['#667eea', '#764ba2', '#f093fb', '#4facfe', '#43e97b']
explode = (0.1, 0, 0, 0, 0)  # Explode the first slice

plt.figure(figsize=(10, 8))
plt.pie(sizes, explode=explode, labels=labels, colors=colors, autopct='%1.1f%%', 
        shadow=True, startangle=90, textprops={'fontsize': 12, 'fontweight': 'bold'})
plt.title('Programming Language Market Share', fontsize=16, fontweight='bold', pad=20)
plt.axis('equal')  # Equal aspect ratio ensures circular pie
plt.tight_layout()
plt.show()
````

````python
# Donut chart
fig, ax = plt.subplots(figsize=(10, 8))
ax.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%',
       startangle=90, pctdistance=0.85, textprops={'fontsize': 11, 'fontweight': 'bold'})

# Draw circle in center to create donut
centre_circle = plt.Circle((0, 0), 0.70, fc='white')
fig.gca().add_artist(centre_circle)

ax.set_title('Market Share - Donut Chart', fontsize=16, fontweight='bold', pad=20)
ax.axis('equal')
plt.tight_layout()
plt.show()
````

# 8.7 Subplots - Multiple Charts
Create multiple charts in one figure for comprehensive analysis.

````python
# Sample data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) + np.cos(x)
y4 = np.sin(x) * np.cos(x)

# Create 2x2 grid of subplots
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Mathematical Functions Visualization', fontsize=18, fontweight='bold', y=1.02)

# Plot 1: Sin(x)
axes[0, 0].plot(x, y1, color='#667eea', linewidth=2)
axes[0, 0].set_title('sin(x)', fontsize=14, fontweight='bold')
axes[0, 0].set_xlabel('x')
axes[0, 0].set_ylabel('y')
axes[0, 0].grid(True, alpha=0.3)
# Plot 2: Cos(x)
axes[0, 1].plot(x, y2, color='#ff6b6b', linewidth=2)
axes[0, 1].set_title('cos(x)', fontsize=14, fontweight='bold')
axes[0, 1].set_xlabel('x')
axes[0, 1].set_ylabel('y')
axes[0, 1].grid(True, alpha=0.3)

# Plot 3: Sin(x) + Cos(x)
axes[1, 0].plot(x, y3, color='#4ecdc4', linewidth=2)
axes[1, 0].set_title('sin(x) + cos(x)', fontsize=14, fontweight='bold')
axes[1, 0].set_xlabel('x')
axes[1, 0].set_ylabel('y')
axes[1, 0].grid(True, alpha=0.3)
# Plot 4: Sin(x) * Cos(x)
axes[1, 1].plot(x, y4, color='#764ba2', linewidth=2)
axes[1, 1].set_title('sin(x) * cos(x)', fontsize=14, fontweight='bold')
axes[1, 1].set_xlabel('x')
axes[1, 1].set_ylabel('y')
axes[1, 1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````
