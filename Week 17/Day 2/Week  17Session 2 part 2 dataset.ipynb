MODULE 7: Pandas - Data Analysis Library

7.1 Introduction to Pandas

Pandas is the most important library for data analysis in Python. It provides:

* DataFrame: 2D labeled data structure (like Excel spreadsheet)
* Series: 1D labeled array
* Data cleaning and preprocessing tools
* Easy handling of missing data
* Reading/writing data from various formats (CSV, Excel, SQL, etc.)
* Powerful grouping and aggregation capabilities

```python
# Import Pandas
import pandas as pd
import numpy as np

# Check version
print(f"Pandas version: {pd.__version__}")
```

# 7.2 Series - 1D Labeled Array

```python
# Creating Series from list
temperatures = pd.Series([23.5, 25.0, 22.8, 24.5, 26.2])
print("Temperature Series:")
print(temperatures)
print(f"\nData type: {temperatures.dtype}")
```

```python
# Series with custom index
temps_labeled = pd.Series([23.5, 25.0, 22.8, 24.5, 26.2],
                          index=['Mon', 'Tue', 'Wed', 'Thu', 'Fri'])
print("\nLabeled Temperature Series:")
print(temps_labeled)
```

```python
# Accessing elements
print(f"\nMonday temperature: {temps_labeled['Mon']}")
print(f"First element (position): {temps_labeled.iloc[0]}")
print(f"\nWed-Fri temperatures:\n{temps_labeled['Wed':'Fri']}")
```

```python
# Series from dictionary
population = pd.Series({
    'New York': 8336817,
    'Los Angeles': 3979576,
    'Chicago': 2693976,
    'Houston': 2320268
})
print("\nCity Population:")
print(population)
# Series operations
print(f"\nMean population: {population.mean():,.0f}")
print(f"Max population: {population.max():,}")
print(f"Total population: {population.sum():,}")
    
```

# 7.3 DataFrame - 2D Labeled Data Structure

DataFrames are the heart of Pandas - think of them as Excel spreadsheets in Python!

```python
# Creating DataFrame from dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [25, 30, 35, 28, 32],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix'],
    'Salary': [70000, 80000, 75000, 85000, 72000]
}
data
```

```python
df = pd.DataFrame(data)
print("Employee DataFrame:")
print(df)
```

```python
df.columns
```

```python
# الحصول على معلومات أساسية
print("\n=== DataFrame Information ===")
print(f"Shape: {df.shape}")           # يعرض عدد الصفوف وعدد الأعمدة
print(f"Columns: {df.columns.tolist()}") # يعرض أسماء الأعمدة الموجودة
print(f"Index: {df.index.tolist()}")   # يعرض الفهارس (عناوين الصفوف)
print(f"\nData types:\n{df.dtypes}")   # يعرض نوع البيانات في كل عمود
```

```python
# First/Last rows
print("\nFirst 3 rows:")
print(df.head(3))

print("\nLast 2 rows:")
print(df.tail(2))
```

```python
# Info and Description
print("\n=== DataFrame Info ===")
df.info()

print("\n=== Statistical Description ===")
print(df.describe())
```

### 7.4 Selecting and Filtering Data

```python
# Continue with previous employee DataFrame

# Select single column (returns Series)
names = df['Name']
print("Names (Series):")
print(names)
print(f"Type: {type(names)}\n")
```

```python
# Select multiple columns (returns DataFrame)
subset = df[['Name', 'Salary']]
print("Name and Salary:")
print(subset)
print(f"Type: {type(subset)}\n")
```

```python
# Select rows by index position - iloc
print("First row (iloc[0]):")
print(df.iloc[0])
print("\nFirst 3 rows (iloc[0:3]):")
print(df.iloc[0:3])
```

```python
# Select rows by label - loc هنا هتجيبي واحدو ثلاثه 
print("\nRows 1 and 3 (loc[[1, 3]]):")
print(df.loc[[1, 3]])
```

```python
# Boolean Filtering
print("\n=== Filtering Examples ===")
d=df['Age'] > 30
print(d)
# Employees older than 30
older_than_30 = df[df['Age'] > 30]
print("Employees older than 30:")
print(older_than_30)
```

```python
# Salary >= 75000
high_earners = df[df['Salary'] >= 75000]
print("\nHigh earners (>= $75,000):")
print(high_earners)
```

```python
 # Multiple conditions (AND - &)
filtered = df[(df['Age'] > 28) & (df['Salary'] > 75000)]
print("\nAge > 28 AND Salary > $75,000:")
print(filtered)
```

```python
# Multiple conditions (OR - |)
filtered_or = df[(df['City'] == 'New York') | (df['City'] == 'Chicago')]
print("\nEmployees in New York OR Chicago:")
print(filtered_or)
```

```python
# Using isin()
cities_of_interest = ['New York', 'Chicago', 'Cairo']
filtered_isin = df[df['City'].isin(cities_of_interest)]
print("\nEmployees in specific cities:")
print(filtered_isin)
```

### 7.5 Adding, Modifying, and Deleting Columns

```python
# Create a copy to work with
df_work = df.copy()
```

```python
# Add new column
df_work['Bonus'] = df_work['Salary'] * 0.10
print("After adding Bonus column:")
print(df_work)
```

```python
# Add column based on condition
df_work['Experience_Level'] = df_work['Age'].apply(lambda x: 'Senior' if x >= 30 else 'Junior')
print("\nAfter adding Experience_Level:")
print(df_work)
```

```python
# Modify existing column
df_work['Salary'] = df_work['Salary'] * 1.05  # 5% raise
print("\nAfter 5% salary increase:")
print(df_work[['Name', 'Salary']])
```

```python
# Rename columns
df_work = df_work.rename(columns={'Salary': 'Annual_Salary', 'Age': 'Years_Old'})
print("\nAfter renaming columns:")
print(df_work.columns.tolist())
df_work
```

```python
# Delete column
df_work = df_work.drop('Bonus', axis=1, inplace=True)
print("\nAfter dropping Bonus column:")
print(df_work.columns.tolist())
```

## 7.6 Handling Missing Data

```python
# Create DataFrame with missing values
data_with_nulls = {
    'Product': ['A', 'B', 'C', 'D', 'E', 'F'],
    'Price': [10.5, np.nan, 15.0, 12.5, np.nan, 18.0],
    'Quantity': [100, 150, np.nan, 200, 175, 120],
    'Category': ['Electronics', 'Electronics', None, 'Clothing', 'Clothing', 'Electronics']
}

df_nulls = pd.DataFrame(data_with_nulls)
print("DataFrame with missing values:")
print(df_nulls)
```

```python
# Detect missing values
print("\nMissing values (True = missing):")
print(df_nulls.isnull())
```

```python
# Detect missing values
print("\nMissing values (True = missing):")
print(df_nulls.isnull().sum())
```
